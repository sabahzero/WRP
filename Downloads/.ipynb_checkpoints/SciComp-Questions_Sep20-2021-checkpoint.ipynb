# Estimation
### How many grad students are studying biology or chemistry in the US right now?
Biology: ~33,434, Chemistry: ~22,289
<br> Total: ~55,723

- Assumption 1: Graduate students = Masters + PhD
- Assumption 2: The number of students who switch from Masters to PhD or PhD to Masters is negligible
- Assumption 3: In the sciences, majority of grad students are doing a PhD (Graduate students = 2*Masters + PhD) 
<br>> Reasoning: PhDs are considered more valuable in Biology and Chemistry than an MSc, versus in Engineering an MSc may be better (ref [Slate article](https://slate.com/technology/2012/08/what-is-the-value-of-a-science-phd-is-graduate-school-worth-the-effort.html), [Harvard article](https://sitn.hms.harvard.edu/flash/2015/what-the-heck-do-you-do-with-a-ph-d-in-the-biomedical-sciences/), and [OEDb article](https://oedb.org/advice/masters-vs-phd-degrees-for-graduate-students/) 
- Assumption 4: Biology is more popular than Chemistry (3 Biology for every 2 Chemistry students)
<br>> Reasoning: Biology is one of the most popular across all undergraduate majors, and the most popular in the sciences (extrapolation from [indeed.com article](https://www.indeed.com/career-advice/career-development/most-popular-majors))

```python
## How many graduate students are there in the United States today? ~3.98 million
### www.census.gov (https://www.census.gov/library/stories/2019/12/school-enrollment-college-down-graduate-school-up.html)

### Create dataframe 
import pandas as pd  # Package to create dataframe of graduate students 

GS = {'Year': ['2011', '2012', '2013', '2014', '2015', '2016', '2017', '2018'], 
      'Million': [3.77, 3.76, 3.73, 3.68, 3.67, 3.70, 3.81, 4.08]}  
gs = pd.DataFrame(GS)  

gs # Print to validate
```

```python
### Predict how many millions of graduate students are in 2021 (census data goes to 2018)
#### https://stackabuse.com/linear-regression-in-python-with-scikit-learn/
from sklearn.model_selection import train_test_split # sklearn package for linear regression (and training)
from sklearn.linear_model import LinearRegression

#### Setting up training data and associated test 
X = gs.iloc[:, :-1].values
y = gs.iloc[:, 1].values
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
regressor = LinearRegression()
regressor.fit(X_train, y_train)

print('b =', regressor.intercept_) # -57.05620000000007
print('m =', regressor.coef_, '\n') # 0.0302

#### Test that prediction matches with actual
y_pred = regressor.predict(X_test) 
gs_test = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(gs_test, '\n')

#### y = mx + b (keep exact number, no rounding, for precision)
print('There are approximately', (0.0302)*2021 - 57.05620000000007, 'graduate students in the US today')
```
