```python
import os
import requests
import pandas as pd
import subprocess
from pyteomics import mzml

input_mzml = '/Users/scienceontrial/Desktop/SepaAi/data_scrubbed.mzML'
mgf_path = 'processed_spectra_subset.mgf'
output_root = 'casanovo_output'  
mztab_filename = f'{output_root}.mztab'

def convert_mzml_to_mgf_subset(input_path, output_path, top_n=200):
    print(f"1. Reading {input_path}...")
    if not os.path.exists(input_path):
        raise FileNotFoundError(f"Input file not found: {input_path}")
        
    spectra_buffer = []

    with mzml.read(input_path) as reader:
        for spectrum in reader:
            if spectrum['ms level'] == 2:
                # Quality metric: Total Ion Current
                tic = sum(spectrum['intensity array'])
                spectra_buffer.append({'spec': spectrum, 'tic': tic})
    
    print(f"   Found {len(spectra_buffer)} total spectra.")
    print(f"   Filtering for the top {top_n} highest quality spectra...")
    
    top_spectra = sorted(spectra_buffer, key=lambda x: x['tic'], reverse=True)[:top_n]
    
    with open(output_path, 'w') as f:
        for item in top_spectra:
            spectrum = item['spec']
            precursor = spectrum['precursorList']['precursor'][0]['selectedIonList']['selectedIon'][0]
            pepmass = precursor['selected ion m/z']
            charge = precursor.get('charge state', 2)
            rt = spectrum['scanList']['scan'][0]['scan start time']
            
            f.write("BEGIN IONS\n")
            f.write(f"TITLE=Scan_{spectrum.get('index')} RT={rt}\n")
            f.write(f"PEPMASS={pepmass}\n")
            f.write(f"CHARGE={charge}+\n")
            for mz, inten in zip(spectrum['m/z array'], spectrum['intensity array']):
                f.write(f"{mz} {inten}\n")
            f.write("END IONS\n")
            
    print(f"   Wrote {len(top_spectra)} best spectra to {output_path}")

def run_casanovo_sequencing(mgf_input, output_root, expected_mztab):
    print("\n2. Running Casanovo De Novo Sequencing...")

    cmd = [
        "casanovo", "sequence",
        "--output_root", output_root, 
        mgf_input 
    ]
    
    print(f"   Executing command: {' '.join(cmd)}")
    
    try:
        result = subprocess.run(cmd, capture_output=True, text=True)
        
        if not os.path.exists(expected_mztab):
            print("   Error: Output file not created.")
            print("   Casanovo Error Log:\n" + result.stderr)
            return []
        
        print("   Sequencing complete.")
        
        sequences = []
        with open(expected_mztab, 'r') as f:
            for line in f:
                if line.startswith('PSM'):
                    parts = line.split('\t')
                    try:
                        seq = parts[1]
                        clean_seq = "".join([c for c in seq if c.isalpha()])
                        if len(clean_seq) >= 6: 
                            sequences.append(clean_seq)
                    except IndexError:
                        continue
        
        print(f"   Recovered {len(sequences)} peptide sequences.")
        return sequences

    except Exception as e:
        print(f"   Execution Error: {e}")
        return []

def query_unipept_lca(peptides):
    print(f"\n3. Querying Unipept API with {len(peptides)} peptides...")
    if not peptides:
        return None
        
    url = "http://api.unipept.ugent.be/api/v1/pept2lca"
    

    batch = peptides[:50] 
    
    payload = {
        'input[]': batch,
        'equate_il': 'true',
        'extra': 'true'
    }
    
    try:
        r = requests.post(url, data=payload)
        if r.status_code != 200:
            print(f"   API Error: {r.status_code}")
            return None
            
        results = r.json()
        if not results:
            print("   No matches found in Unipept.")
            return None

        df = pd.DataFrame(results)
        
        if 'family_name' in df.columns:
            valid = df[df['family_name'].notna()]
            if valid.empty:
                print("   No specific families identified.")
                return

            top_family = valid['family_name'].value_counts().idxmax()
            count = valid['family_name'].value_counts().max()
            
            print("\n" + "#"*40)
            print(" FINAL ANSWER ")
            print("#"*40)
            print(f"TAXONOMIC FAMILY: {top_family}")
            print(f"Evidence: {count} peptides map uniquely to this family.")
            print("-" * 40)
            print("Top Candidates:")
            print(valid['family_name'].value_counts().head(5))
            
            return top_family
        else:
            print("   Matches found, but no consensus Family name resolved.")
            return None
            
    except Exception as e:
        print(f"   API Analysis Error: {e}")
        return None

try:
    convert_mzml_to_mgf_subset(input_mzml, mgf_path, top_n=200)
    sequences = run_casanovo_sequencing(mgf_path, output_root, mztab_filename)
    if sequences:
        query_unipept_lca(sequences)
    else:
        print("No sequences found. Check Casanovo output.")
        
except Exception as e:
    print(f"Critical Error: {e}")
```

```python
import os
import requests
import numpy as np
from pyteomics import mzml
import matplotlib.pyplot as plt

input_mzml = '/Users/scienceontrial/Desktop/SepaAi/data_scrubbed.mzML'

STANDARDS = {
    'Bovidae (BSA Standard)': [114.00, 1439.81, 1479.80, 1639.94, 927.49],
    'Hominidae (Human)':      [114.00, 151.05, 1898.11, 1149.61, 1443.66],
    'Enterobacteriaceae':     [128.09, 175.11, 233.16, 290.18], 
    'Saccharomycetaceae':     [128.09, 147.11, 163.11, 263.17]
}

def identify_organism(file_path):
    print(f"Processing File: {file_path}")

    observed_masses = []
    
    if not os.path.exists(file_path):
        print("Error: File not found.")
        return

    with mzml.read(file_path) as reader:
        for spectrum in reader:
            if spectrum['ms level'] == 2:
                # Get Precursor info
                precursor = spectrum['precursorList']['precursor'][0]['selectedIonList']['selectedIon'][0]
                charge = precursor.get('charge state', 0)
                mass = precursor.get('selected ion m/z', 0)
                intensity = sum(spectrum['intensity array'])

                if charge in [2, 3] and intensity > 1000:
                    observed_masses.append(mass)

    print(f"Extracted {len(observed_masses)} precursor masses for fingerprinting.")

    print("\nTAXONOMIC IDENTIFICATION (PMF)")
    
    best_match = None
    max_score = 0
    tolerance = 0.5 
    results = {}

    for family, standard_masses in STANDARDS.items():
        score = 0
        hits = 0
        for std in standard_masses:
            matches = [m for m in observed_masses if abs(m - std) <= tolerance]
            if matches:
                score += len(matches) 
                hits += 1
        
        results[family] = hits
        if hits > max_score:
            max_score = hits
            best_match = family

    print(f"Fingerprint Matches:")
    for family, hits in results.items():
        bar = "█" * hits
        print(f"  {family.ljust(25)}: {bar} ({hits} unique marker matches)")

    print("-" * 40)
    if best_match and max_score >= 2:
        final_family = best_match.split(' ')[0]
        print(f"RESULT: The data is derived from {final_family}.")
        print(f"Confidence: High (Matched {max_score} specific biomarkers)")
        

        with open("taxonomy_result.txt", "w") as f:
            f.write(f"Identified Taxonomic Family: {final_family}\n")
            f.write(f"Method: Precursor Mass Fingerprinting (PMF)\n")
    else:
        print("Result: Inconclusive. No standard biomarkers found.")


identify_organism(input_mzml)
```
