### BLAST hits correspond to fungal ITS regions

### Count the most common 5′ start sequences to identify candidate primers.

```python
from collections import Counter

k = 35  # look at the first 35 bp of each read (approx. primer length)

start_counts = Counter()

for rec in records:
    seq = str(rec.seq)
    if len(seq) >= k:
        start_counts[seq[:k]] += 1

start_counts.most_common(10)
```

### show the most common 3′-end motifs

```python
from collections import Counter

k = 35  # same window size for the 3' end

end_counts = Counter()

for rec in records:
    seq = str(rec.seq)
    if len(seq) >= k:
        end_counts[seq[-k:]] += 1

end_counts.most_common(10)
```

```python
from collections import Counter

# get the top 20 most common 5' motifs
top_start = [seq for seq, count in start_counts.most_common(20)]

def hamming(a, b):
    if len(a) != len(b):
        return None
    return sum(x != y for x, y in zip(a, b))

# Group sequences that differ by at most 2 mismatches
groups = []
visited = set()

for i, s in enumerate(top_start):
    if s in visited:
        continue
    group = [s]
    visited.add(s)
    for t in top_start[i+1:]:
        if t in visited:
            continue
        dist = hamming(s, t)
        if dist is not None and dist <= 2:
            group.append(t)
            visited.add(t)
    groups.append(group)

from collections import Counter

def consensus(seqs):
    if not seqs:
        return ""
    length = len(seqs[0])
    cons = []
    for i in range(length):
        col = [s[i] for s in seqs]
        cons.append(Counter(col).most_common(1)[0][0])
    return "".join(cons)

for i, g in enumerate(groups, 1):
    print(f"Group {i} (n={len(g)}):")
    for s in g:
        print(" ", s)
    print("Consensus:", consensus(g))
```

```python
# Reuse 'groups' and 'start_counts' from before

primer_stats = []

for g in groups:
    total = sum(start_counts[s] for s in g)
    primer_stats.append({
        "consensus": g[0],   
        "n_sequences": len(g),
        "total_reads": total
    })

# Sort by how many reads they account for
primer_stats = sorted(primer_stats, key=lambda x: x["total_reads"], reverse=True)

for p in primer_stats[:10]:
    print(f"Consensus: {p['consensus']}")
    print(f"  Number of similar 5' motifs in group: {p['n_sequences']}")
    print(f"  Total reads starting with this group: {p['total_reads']}")
```

```python
from Bio.Seq import Seq

def hamming(a, b):
    if len(a) != len(b):
        return None
    return sum(x != y for x, y in zip(a, b))

def try_match(candidate, known):
    # match candidate to known (exact length match)
    if len(candidate) >= len(known):
        window = candidate[:len(known)]
        return hamming(window, known)
    return None
```

### ITS4 is present exactly at the start of Candidate 1 → that's one of the amplicon primers

```python
from Bio.Seq import Seq

candidate2 = "GTGAATCATCGAATCTTTGAACGCACATTGCGCCC"

def best_window_distance(candidate, primer):
    """Slide the primer along the candidate and return the best (min) Hamming distance."""
    if len(candidate) < len(primer):
        return None
    best = None
    for i in range(len(candidate) - len(primer) + 1):
        window = candidate[i:i+len(primer)]
        d = hamming(window, primer)
        if best is None or d < best:
            best = d
    return best

print("Candidate 2:", candidate2, "\n")

for name, pseq in known_primers.items():
    rc_pseq = str(Seq(pseq).reverse_complement())
    d_fwd = best_window_distance(candidate2, pseq)
    d_rc  = best_window_distance(candidate2, rc_pseq)
    print(f"{name}: best distance vs primer = {d_fwd}, vs RC = {d_rc}")
```

### The second frequent 5' motif does not clearly match common 16S/ITS primers

```python
from Bio.Seq import Seq

def approx_contains(read_seq, primer_seq, max_mismatches=2):
    """
    Return True if primer_seq occurs in read_seq with <= max_mismatches
    in any window of the same length.
    """
    Lr = len(read_seq)
    Lp = len(primer_seq)
    if Lr < Lp:
        return False
    for i in range(Lr - Lp + 1):
        window = read_seq[i:i+Lp]
        mismatches = sum(a != b for a, b in zip(window, primer_seq))
        if mismatches <= max_mismatches:
            return True
    return False

primer_hits = {
    name: {"primer_seq": pseq, "forward_hits": 0, "reverse_hits": 0}
    for name, pseq in known_primers.items()
}

for rec in records:
    s = str(rec.seq)
    for name, pseq in known_primers.items():
        rc = str(Seq(pseq).reverse_complement())
        if approx_contains(s, pseq, max_mismatches=2):
            primer_hits[name]["forward_hits"] += 1
        if approx_contains(s, rc, max_mismatches=2):
            primer_hits[name]["reverse_hits"] += 1

for name, stats in primer_hits.items():
    print(
        f"{name}: forward_hits={stats['forward_hits']}, "
        f"reverse_hits={stats['reverse_hits']}"
    )
```

```python
from Bio.Seq import Seq

def best_match_pos(read_seq, primer_seq):
    """Return (best_position, mismatches) of primer_seq in read_seq."""
    Lr = len(read_seq)
    Lp = len(primer_seq)
    best_pos = None
    best_mismatches = None
    if Lr < Lp:
        return None, None
    for i in range(Lr - Lp + 1):
        window = read_seq[i:i+Lp]
        mismatches = sum(a != b for a, b in zip(window, primer_seq))
        if best_mismatches is None or mismatches < best_mismatches:
            best_mismatches = mismatches
            best_pos = i
    return best_pos, best_mismatches

its4 = known_primers["ITS4"]
f515 = known_primers["515F"]

print("Checking ITS4 and 515F positions in reads:\n")

for rec in records[:10]:  # just first 10 reads as example
    s = str(rec.seq)
    pos_its4, mis_its4 = best_match_pos(s, its4)
    pos_515, mis_515 = best_match_pos(s, f515)
    print(f"Read {rec.id}:")
    print(f"  ITS4  -> position {pos_its4}, mismatches {mis_its4}")
    print(f"  515F  -> position {pos_515}, mismatches {mis_515}")
```

```python
known_primers = {
    # 16S primers
    "27F":  "AGAGTTTGATCMTGGCTCAG",
    "338F": "ACTCCTACGGGAGGCAGC",
    "341F": "CCTACGGGNGGCWGCAG",
    "515F": "GTGCCAGCMGCCGCGGTAA",
    "519R": "GWATTACCGCGGCKGCTG",
    "785R": "GACTACHVGGGTATCTAATCC",
    "806R": "GGACTACHVGGGTWTCTAAT",
    "907R": "CCGTCAATTCMTTTRAGTTT",
    "926R": "CCGYCAATTYMTTTRAGTTT",
    "1492R": "TACGGYTACCTTGTTACGACTT",

    # ITS primers
    "ITS1F": "CTTGGTCATTTAGAGGAAGTAA",
    "ITS2":  "GCTGCGTTCTTCATCGATGC",
    "ITS3":  "GCATCGATGAAGAACGCAGC",
    "ITS4":  "TCCTCCGCTTATTGATATGC",
    "ITS86F": "GTGAATCATCGAATCTTTGAA",
}
```

```python
from Bio.Seq import Seq

def approx_contains(read_seq, primer_seq, max_mismatches=2):
    """
    Return True if primer_seq occurs in read_seq with <= max_mismatches
    anywhere.
    """
    Lr = len(read_seq)
    Lp = len(primer_seq)
    if Lr < Lp:
        return False
    for i in range(Lr - Lp + 1):
        window = read_seq[i:i+Lp]
        mismatches = sum(a != b for a, b in zip(window, primer_seq))
        if mismatches <= max_mismatches:
            return True
    return False

primer_hits = {
    name: {"primer_seq": pseq, "forward_hits": 0, "reverse_hits": 0}
    for name, pseq in known_primers.items()
}

for rec in records:
    s = str(rec.seq)
    for name, pseq in known_primers.items():
        rc = str(Seq(pseq).reverse_complement())
        if approx_contains(s, pseq, max_mismatches=2):
            primer_hits[name]["forward_hits"] += 1
        if approx_contains(s, rc, max_mismatches=2):
            primer_hits[name]["reverse_hits"] += 1

for name, stats in primer_hits.items():
    print(
        f"{name}: forward_hits={stats['forward_hits']}, "
        f"reverse_hits={stats['reverse_hits']}"
    )
```

```python
def best_match_pos(read_seq, primer_seq):
    """Return (best_position, mismatches) of primer_seq in read_seq."""
    Lr = len(read_seq)
    Lp = len(primer_seq)
    if Lr < Lp:
        return None, None
    best_pos = None
    best_mismatches = None
    for i in range(Lr - Lp + 1):
        window = read_seq[i:i+Lp]
        mismatches = sum(a != b for a, b in zip(window, primer_seq))
        if best_mismatches is None or mismatches < best_mismatches:
            best_mismatches = mismatches
            best_pos = i
    return best_pos, best_mismatches

its4 = known_primers["ITS4"]

for rec in records[:10]:
    s = str(rec.seq)
    pos, mis = best_match_pos(s, its4)
    print(f"{rec.id}: ITS4 best position = {pos}, mismatches = {mis}")
```
