```python
fasta_file = 'hb010-sequences.fasta'  # Update file path if needed

with open(fasta_file, 'r') as f:
    header = None
    sequence = ''
    count = 0
    for line in f:
        line = line.strip()
        if line.startswith('>'):
            if header:
                print(f"Header: {header}")
                print(f"Sequence start: {sequence[:50]}\n")
                count += 1
                if count >= 5:
                    break
            header = line
            sequence = ''
        else:
            sequence += line

    # Print last sequence
    if header and count < 5:
        print(f"Header: {header}")
        print(f"Sequence start: {sequence[:50]}")
```

```python
# Define common 16S rRNA primers (forward and reverse) frequently used for microbial community profiling
primers = {
    '341F': 'CCTACGGGNGGCWGCAG',
    '515F': 'GTGCCAGCMGCCGCGGTAA',
    '806R': 'GGACTACHVGGGTWTCTAAT',
    '805R': 'GACTACHVGGGTATCTAATCC',
    '27F': 'AGAGTTTGATCMTGGCTCAG',
    '1492R': 'TACGGYTACCTTGTTACGACTT',
    'ITS4': 'TCCTCCGCTTATTGATATGC'
}

# Your sequences (add more if needed)
sequences = {
    'a': 'TCCTCCGCTTATTGATATGCTTAAGTTCAGCGGGTAGTCCTACCTGATTT',
    'b': 'GTGAATCATCGAATCTTTGAACGCACCTTGCGCTCCTTGGTATTCCTTGG',
    'c': 'TCCTCCGCTTATTGATATGCTTAAGTTCAGCGGGTATTCCTACCTGATCC',
    'd': 'GTGAATCATCGAATCTTTGAACGCACATTGCGCCCGTTAGCATTCTAGCG',
    'e': 'TCCTCCGCTTATTGATATGCTTAAGTTCAGCGGGTATTCCTACCTGATCC'
}

from difflib import SequenceMatcher

def similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()

# Function to find best primer match for each sequence
for seq_id, seq in sequences.items():
    best_match = None
    best_score = 0
    primer_matched_seq = ''
    for primer_name, primer_seq in primers.items():
        # Compare the first len(primer_seq) bases of seq to primer sequence (ignoring Ns/Ws via simple filtering can be done but here exact bases only)
        seq_substr = seq[:len(primer_seq)].upper()
        # For ambiguous bases, replace to allow simple matching, here we just handle N->any by ignoring exact ambiguous matching for simplicity
        # Calculate similarity ratio
        score = similarity(seq_substr, primer_seq)
        if score > best_score:
            best_score = score
            best_match = primer_name
            primer_matched_seq = seq_substr
    print(f"Sequence {seq_id} best matches primer {best_match} with score {best_score:.2f} (sequence start: {primer_matched_seq})")
```
