```python
import os
import glob
import pandas as pd
import numpy as np
from scipy.stats import spearmanr, pearsonr, zscore
import warnings
warnings.filterwarnings('ignore')

print("Loading RNA-seq data...")
rna_df = pd.read_csv('transcriptomics_anon.cct', sep='\t', index_col=0)
print(f"RNA-seq: {rna_df.shape[0]} genes × {rna_df.shape[1]} samples")

print("\nLoading Proteome data...")
protein_df = pd.read_csv('proteomics_anon.cct', sep='\t', index_col=0)
print(f"Proteome: {protein_df.shape[0]} proteins × {protein_df.shape[1]} samples")

print("\n Data loaded successfully")
```

```python
rna_samples = set(rna_df.columns)
protein_samples = set(protein_df.columns)
common_samples = sorted(rna_samples.intersection(protein_samples))


print("STEP 1: Sample Pairing Analysis")


print(f"\nRNA samples: {len(rna_samples)}")
print(f"Protein samples: {len(protein_samples)}")
print(f"Common samples (paired): {len(common_samples)}")

# Check for mismatches
only_rna = rna_samples - protein_samples
only_protein = protein_samples - rna_samples

if only_rna:
    print(f"\n Samples only in RNA: {len(only_rna)}")
    print(f"  {list(only_rna)[:5]}")  # Show first 5

if only_protein:
    print(f"\n⚠ Samples only in Protein: {len(only_protein)}")
    print(f"  {list(only_protein)[:5]}")  # Show first 5

if len(common_samples) == len(rna_samples) == len(protein_samples):
    print("\n PERFECT PAIRING: All sample IDs match between datasets")
else:
    print("\n IMPERFECT PAIRING: Some samples are dataset-specific")

print(f"\nFirst 10 paired samples: {common_samples[:10]}")
```

```python
# Define metabolic pathway genes. I have obtained from a literature search
metabolic_genes = {
    'Glycolysis': [
        'HK1', 'HK2', 'HK3', 'GCK', 'GPI', 'PFKL', 'PFKM', 'PFKP',
        'ALDOA', 'ALDOB', 'ALDOC', 'TPI1', 'GAPDH', 'GAPDHS',
        'PGK1', 'PGK2', 'PGAM1', 'PGAM2', 'ENO1', 'ENO2', 'ENO3',
        'PKM', 'PKLR', 'LDHA', 'LDHB', 'LDHC'
    ],
    'TCA_cycle': [
        'CS', 'ACO1', 'ACO2', 'IDH1', 'IDH2', 'IDH3A', 'IDH3B', 'IDH3G',
        'OGDH', 'OGDHL', 'DLST', 'SUCLA2', 'SUCLG1', 'SUCLG2',
        'SDHA', 'SDHB', 'SDHC', 'SDHD', 'FH', 'MDH1', 'MDH2'
    ],
    'OXPHOS': [
        'NDUFA1', 'NDUFA2', 'NDUFA3', 'NDUFA4', 'NDUFA5', 'NDUFA6',
        'NDUFA7', 'NDUFA8', 'NDUFA9', 'NDUFA10', 'NDUFA11', 'NDUFA12', 'NDUFA13',
        'NDUFB1', 'NDUFB2', 'NDUFB3', 'NDUFB4', 'NDUFB5', 'NDUFB6',
        'NDUFB7', 'NDUFB8', 'NDUFB9', 'NDUFB10', 'NDUFB11',
        'NDUFS1', 'NDUFS2', 'NDUFS3', 'NDUFS4', 'NDUFS5', 'NDUFS6',
        'NDUFS7', 'NDUFS8', 'NDUFV1', 'NDUFV2', 'NDUFV3',
        'UQCRC1', 'UQCRC2', 'UQCRFS1', 'UQCRB', 'UQCRH', 'UQCRQ', 'CYC1',
        'COX4I1', 'COX4I2', 'COX5A', 'COX5B', 'COX6A1', 'COX6A2',
        'COX6B1', 'COX6B2', 'COX6C', 'COX7A1', 'COX7A2', 'COX7B', 'COX7C',
        'COX8A', 'COX10', 'COX11', 'COX15', 'COX17',
        'ATP5F1A', 'ATP5F1B', 'ATP5F1C', 'ATP5F1D', 'ATP5F1E',
        'ATP5PB', 'ATP5PD', 'ATP5PF', 'ATP5PO', 'ATP5ME', 'ATP5MF', 'ATP5MG'
    ]
}

# Flatten to get all unique metabolic genes
all_metabolic_genes = []
for pathway, genes in metabolic_genes.items():
    all_metabolic_genes.extend(genes)
all_metabolic_genes = list(set(all_metabolic_genes))

# Check coverage
rna_genes = set(rna_df.index)
protein_genes = set(protein_df.index)
metabolic_in_data = [g for g in all_metabolic_genes if g in rna_genes and g in protein_genes]

print(f"Total unique metabolic genes defined: {len(all_metabolic_genes)}")
print(f"  - Glycolysis: {len(metabolic_genes['Glycolysis'])} genes")
print(f"  - TCA cycle: {len(metabolic_genes['TCA_cycle'])} genes")
print(f"  - OXPHOS: {len(metabolic_genes['OXPHOS'])} genes")

print(f"\nMetabolic genes with both RNA + Protein data: {len(metabolic_in_data)}")

for pathway, genes in metabolic_genes.items():
    available = [g for g in genes if g in metabolic_in_data]
    print(f"  - {pathway}: {len(available)}/{len(genes)} available")
```

```python
# Calculate discordance for each sample

print("STEP 2: Calculate Per-Sample Metabolic Discordance")


sample_scores = []

for sample in common_samples:
    # Extract metabolic gene values for this sample
    rna_vals = rna_df.loc[metabolic_in_data, sample]
    protein_vals = protein_df.loc[metabolic_in_data, sample]
    
    # Remove NA values (must be present in both)
    valid_mask = (~rna_vals.isna()) & (~protein_vals.isna())
    
    if valid_mask.sum() < 10:  # Need at least 10 genes
        continue
    
    rna_clean = rna_vals[valid_mask].astype(float).values
    protein_clean = protein_vals[valid_mask].astype(float).values
    
    # Calculate Spearman correlation (rank-based, robust to outliers)
    corr, pval = spearmanr(rna_clean, protein_clean)
    
    # Calculate mean absolute residual
    coeffs = np.polyfit(rna_clean, protein_clean, 1)
    predicted = coeffs[0] * rna_clean + coeffs[1]
    residuals = np.abs(protein_clean - predicted)
    mean_residual = np.mean(residuals)
    
    # Discordance score = 1 - correlation (higher = more discordant)
    discordance_score = 1 - corr
    
    sample_scores.append({
        'sample_id': sample,
        'n_genes': valid_mask.sum(),
        'spearman_corr': corr,
        'discordance_score': discordance_score,
        'mean_residual': mean_residual
    })

scores_df = pd.DataFrame(sample_scores)

# Add Z-scores
scores_df['discordance_zscore'] = zscore(scores_df['discordance_score'])

print(f"\nAnalyzed {len(scores_df)} samples")
print(f"\nDiscordance score statistics:")
print(f"  Mean: {scores_df['discordance_score'].mean():.3f}")
print(f"  Median: {scores_df['discordance_score'].median():.3f}")
print(f"  Std Dev: {scores_df['discordance_score'].std():.3f}")
print(f"  Min: {scores_df['discordance_score'].min():.3f}")
print(f"  Max: {scores_df['discordance_score'].max():.3f}")

print("\n Discordance scores calculated")
```

```python
# Define threshold for significant discordance
threshold_zscore = 1.5  # Top ~7% most discordant

discordant_samples = scores_df[scores_df['discordance_zscore'] > threshold_zscore].copy()
discordant_samples = discordant_samples.sort_values('discordance_score', ascending=False)


print("STEP 3: Identify Samples with Significant Discordance")


print(f"\nThreshold: Z-score > {threshold_zscore}")
print(f"Samples with SIGNIFICANT DISCORDANCE: {len(discordant_samples)}")

print(f"\n{'Sample ID':<15} {'Spearman r':<12} {'Discordance':<12} {'Z-score':<10} {'#Genes':<8}")
print("-"*60)
for _, row in discordant_samples.iterrows():
    print(f"{row['sample_id']:<15} {row['spearman_corr']:>11.3f} {row['discordance_score']:>11.3f} {row['discordance_zscore']:>9.2f} {row['n_genes']:>7}")
```

```python
# Analyze which genes show most discordance

print("STEP 4: Gene-Level Discordance Analysis")


gene_correlations = []

for pathway, genes in metabolic_genes.items():
    for gene in genes:
        if gene not in metabolic_in_data:
            continue
        
        rna_vals = rna_df.loc[gene, common_samples]
        protein_vals = protein_df.loc[gene, common_samples]
        
        valid_mask = (~rna_vals.isna()) & (~protein_vals.isna())
        
        if valid_mask.sum() < 10:
            continue
        
        rna_clean = rna_vals[valid_mask].astype(float)
        protein_clean = protein_vals[valid_mask].astype(float)
        
        corr, pval = pearsonr(rna_clean, protein_clean)
        
        gene_correlations.append({
            'gene': gene,
            'pathway': pathway,
            'correlation': corr,
            'p_value': pval,
            'n_samples': valid_mask.sum()
        })

gene_corr_df = pd.DataFrame(gene_correlations)
gene_corr_df = gene_corr_df.sort_values('correlation')

print(f"\nAnalyzed {len(gene_corr_df)} metabolic genes")

print(f"\nTop 10 MOST DISCORDANT GENES (lowest/negative correlation):")
print(gene_corr_df.head(10)[['gene', 'pathway', 'correlation', 'p_value']].to_string(index=False))

print(f"\nTop 10 MOST CONCORDANT GENES (highest correlation):")
print(gene_corr_df.tail(10)[['gene', 'pathway', 'correlation', 'p_value']].to_string(index=False))
```

```python

print(f"\n1. SAMPLE PAIRING:")
print(f"   {len(common_samples)} samples correctly paired")
print(f"   All sample IDs match between RNA-seq and proteomics datasets")

print(f"\n2. SAMPLES WITH SIGNIFICANT METABOLIC DISCORDANCE:")
print(f"   {len(discordant_samples)} samples identified (Z-score > {threshold_zscore})")
print(f"\n   Sample IDs:")
for _, row in discordant_samples.iterrows():
    print(f"   - {row['sample_id']}: Spearman r={row['spearman_corr']:.3f}, Z={row['discordance_zscore']:.2f}")

print(f"\n3. KEY METABOLIC GENES DRIVING DISCORDANCE:")
top_genes = gene_corr_df.head(5)
for _, row in top_genes.iterrows():
    print(f"   - {row['gene']} ({row['pathway']}): r={row['correlation']:.3f}")

# Save results
scores_df.to_csv('sample_discordance_scores.csv', index=False)
discordant_samples.to_csv('discordant_samples.csv', index=False)
gene_corr_df.to_csv('gene_level_correlations.csv', index=False)

```
