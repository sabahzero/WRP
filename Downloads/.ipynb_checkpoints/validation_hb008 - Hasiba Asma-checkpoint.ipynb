```python
############################################
# MysteryBench: Tissue Identification Validator
############################################

!pip install mygene
import pandas as pd
import mygene

###############################
# STEP 1 — Load raw sample file
###############################

df = pd.read_csv("sample1.tsv",
                 sep="\t", comment="#")
df["ID_REF"] = df["ID_REF"].astype(str)  # ensure string keys

print("Raw table loaded:")
display(df.head())


###############################################
# STEP 2 — Map Entrez IDs → Gene Symbols (MyGene)
###############################################

mg = mygene.MyGeneInfo()

mapping = mg.querymany(
    df["ID_REF"].tolist(),
    scopes="entrezgene",
    fields="symbol",
    species="human"
)

map_df = pd.DataFrame(mapping)[["query", "symbol"]].rename(
    columns={"query": "ID_REF"}
)
map_df["ID_REF"] = map_df["ID_REF"].astype(str)

###############################################
# STEP 3 — Merge expression data with gene symbols
###############################################

df2 = df.merge(map_df, on="ID_REF", how="left")

print("Mapped table (with gene symbols):")
display(df2.head())

###############################################
# STEP 4 — Save MAPPED version (this was missing!)
###############################################

df2.to_csv("sample1_mapped.tsv", sep="\t", index=False)
print("\nSaved file: sample1_mapped.tsv")

########################################################
# STEP 5 — RELOAD mapped table for the mystery question
########################################################

df = pd.read_csv("sample1_mapped.tsv", sep="\t")
print("\nReloaded mapped file:")
display(df.head())


###############################################
# STEP 6 — Define marker sets
###############################################

fusiform_markers = ["GFAP", "MBP", "SNAP25", "MAP2", "SYT1"]

liver_markers = ["ALB", "FGA", "FGB", "FGL1", "HP"]
heart_markers = ["TNNT2", "MYH6", "MYH7", "ACTC1", "NKX2-5"]
kidney_markers = ["SLC12A1", "UMOD", "AQP2", "NPHS1", "NPHS2"]
lung_markers = ["SFTPA1", "SFTPB", "SFTPC", "AGER", "SCGB1A1"]
muscle_markers = ["MYOD1", "MYOG", "DES", "ACTN2", "TNNC1"]
pancreas_markers = ["INS", "GCG", "SST", "PPY", "MAFA"]
intestine_markers = ["VIL1", "MUC2", "SI", "FABP2", "LCT"]

tissue_markers = {
    "Brain": fusiform_markers,
    "Liver": liver_markers,
    "Heart": heart_markers,
    "Kidney": kidney_markers,
    "Lung": lung_markers,
    "Muscle": muscle_markers,
    "Pancreas": pancreas_markers,
    "Intestine": intestine_markers
}


###############################################
# STEP 7 — Compute tissue scores
###############################################

scores = {}
for tissue, markers in tissue_markers.items():
    score = df[df["symbol"].isin(markers)]["VALUE"].sum()
    scores[tissue] = score

print("\nTissue marker scores (higher = more likely):")
for tissue, score in scores.items():
    print(f"{tissue}: {score:.2f}")

predicted_tissue = max(scores, key=scores.get)
print(f"\nPredicted tissue: {predicted_tissue}")


###############################################
# STEP 8 — Validate (ground truth known)
###############################################

#expected_tissue = "Fusiform gyrus"
#assert predicted_tissue == expected_tissue, "Validation failed!"
expected_labels = ["Brain", "Fusiform gyrus"]

assert predicted_tissue in expected_labels, \
    f"Validation failed: predicted {predicted_tissue}, expected one of {expected_labels}"

print("\n🎉 Validation successful — prediction matches ground-truth tissue!")
```
