```python
# No. 3 Question
## A college conducts a common test for all the students.
## For the Mathematics portion of this test, the scores are normally distributed 
## with a mean of 502 and a standard deviation of 115. Based on this information, 
## what percentage of students would be expected to score between 400 and 590?

mean = 502
std = 115

### What is the z-score, why do we need it here, and when are other times to use it?
#### https://stats.stackexchange.com/questions/472391/how-to-calculate-z-score-from-mean-and-standard-deviation-in-python

import numpy as np # Why do we need the numpy package, what functions?
## What is the difference between a package vs a library, and why use them?

#### We create a function that calculates z score
def z_score(value, mean, std):
    return (value - mean) / std # How to calculate z-score

z400 = z_score(400, 502, 115) # -0.8869
z590 = z_score(590, 502, 115) # 0.7652
print(z400, z590) # prints out answers from both

#### How would you solve the rest of this problem on your own, using code?
```

```python
# No. 5 Question
## Suppose X is a continuous random variable 
## that is uniformly distributed between 3 and 8. 
## Which one of the following functions can be used to find P(X<=5.8)?

import scipy.stats as st # Why do we need the scipy.stats subpackage, what functions?
dir(st) # directly of all functions in a package, this is under uniform
## https://docs.scipy.org/doc/scipy-0.13.0/reference/generated/scipy.stats.uniform.html

incorrect = 1-st.uniform.cdf(5.8,loc=3,scale=8) # 0.65 > 5.8
correct = st.uniform.cdf(5.8,loc=3,scale=5) # 0.56 < 5.8
print(incorrect, correct)
# Try the rest of the functions in the answers, what do you get? 
```
