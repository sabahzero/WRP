```r
if (!require("BiocManager", quietly = TRUE))
  install.packages("BiocManager")

BiocManager::install("rBLAST")
library(rBLAST)


url <- "https://ftp.ncbi.nlm.nih.gov/genomes/genbank/plant/Brachypodium_distachyon/reference/GCA_000005505.4_Brachypodium_distachyon_v3.0/GCA_000005505.4_Brachypodium_distachyon_v3.0_cds_from_genomic.fna.gz"

dest <- "GCA_000005505.4_Brachypodium_distachyon_v3.0_cds_from_genomic.fna.gz"

download.file(url, destfile = dest, mode = "wb")

system('gunzip GCA_000005505.4_Brachypodium_distachyon_v3.0_cds_from_genomic.fna.gz')
genome= readDNAStringSet('GCA_000005505.4_Brachypodium_distachyon_v3.0_cds_from_genomic.fna')


query=readDNAStringSet('B.distachyon_DEG_scrubbed.fna')

reference= 'GCA_000005505.4_Brachypodium_distachyon_v3.0_cds_from_genomic.fna'

makeblastdb(reference, dbtype='nucl')

bl <- blast(db=reference, type = 'blastn')
cl <- predict(bl, query)

##returns top blast hits with gene IDs 
##gene IDs linked to annotations associated with heat stress 
```
