# Taxonomy Family Prediction with Mass Spectrometry Data

1. Extract Top High Intensity MS2 Spectra data
2. Download Comet/Swissprot and Perform Peptide/Protein Matching
3. Match Proteins/Peptides with Unipept, Obtaining LCA/Counts
4. Find Family From Taxonomy Counts, Species --> Family Count Cumulative

## Installations and Imports

```python
!pip -q install pyteomics ete3 lxml requests

import re
import requests
from google.colab import drive

from pyteomics import mzml
from pyteomics import pepxml

import numpy as np
from ete3 import NCBITaxa
import pandas as pd
```

## Set-Up Paths

```python
drive.mount('/content/drive')
mzml_path = "/content/drive/MyDrive/hb046/data_scrubbed.mzML"
mgf_file_path = '/content/filtered_ms2.mgf'
```

## Obtain Top MS2 Spectra

```python
spectra = []
with mzml.MzML(mzml_path) as reader:
    for spec in reader:
        if spec.get("ms level") != 2 or len(spec.get("m/z array", [])) < 10:
            continue

        # Get precursor info
        try:
            prec = spec["precursorList"]["precursor"][0]["selectedIonList"]["selectedIon"][0]
            mz = float(prec.get("selected ion m/z", 0))
            charge = int(prec.get("charge state", 0))
            rt = float(spec["scanList"]["scan"][0].get("scan start time", 0))
        except:
            continue

        if charge <= 0:  # Skip invalid charges
            continue

        spectra.append({
            'mz': np.array(spec["m/z array"]),
            'intensity': np.array(spec["intensity array"]),
            'precursor_mz': mz,
            'charge': charge,
            'rt': rt
        })

print(f"Total extracted MS2 spectra: "+str(len(spectra)))

# Keep top 10,000 by total intensity
tics = [s['intensity'].sum() for s in spectra]
top_idx = np.argsort(tics)[::-1][:10000]
filtered = [spectra[i] for i in top_idx]

# Write MGF file
with open(mgf_file_path, "w") as f:
    for i, s in enumerate(filtered):
        f.write(f"BEGIN IONS\nTITLE=spec_{i}\n")
        f.write(f"PEPMASS={s['precursor_mz']:.6f}\n")
        f.write(f"CHARGE={s['charge']}+\n")
        f.write(f"RTINSECONDS={s['rt']*60:.2f}\n")
        for m, inten in zip(s['mz'], s['intensity']):
            f.write(f"{m:.5f} {inten:.1f}\n")
        f.write("END IONS\n\n")

print("Total spectra written to MGF: "+str(len(filtered)))
```

## Run Comet on MS2 Spectra

```python
!./comet.linux.exe -p  # Generate params file
!mv comet.params.new comet.params

with open('comet.params', 'r') as f:
    params = f.read()

# Update parameters for Swiss-Prot database
params = re.sub(r'database_name\s*=.*', 'database_name = uniprot_sprot.fasta', params)
params = re.sub(r'decoy_search\s*=.*', 'decoy_search = 1', params)
params = re.sub(r'peptide_mass_tolerance\s*=.*', 'peptide_mass_tolerance = 20.0', params)
params = re.sub(r'peptide_mass_units\s*=.*', 'peptide_mass_units = 2', params)
params = re.sub(r'num_threads\s*=.*', 'num_threads = 0', params)

with open('comet.params', 'w') as f:
    f.write(params)

print("Running database search (this may take 5-10 minutes)...")
!./comet.linux.exe /content/filtered_ms2.mgf
```

## Parse Comet Results For Unique Peptides/Proteins

```python
results = []
with pepxml.read('filtered_ms2.pep.xml') as reader:
    for psm in reader:
        if 'search_hit' in psm and len(psm['search_hit']) > 0:
            hit = psm['search_hit'][0]
            results.append({
                'peptide': hit.get('peptide', 'N/A'),
                'protein': hit.get('protein', 'N/A'),
                'charge': psm.get('assumed_charge', 'N/A'),
            })

pep_prot_df = pd.DataFrame(results)

print('Total peptide-spectrum matches: '+str(len(pep_prot_df)))
print('Total Unique Peptides: '+str(len(pep_prot_df)))
```

## Obtain Taxonomic Classifications with Unipept

```python

peptide_list = pep_prot_df['peptide'].unique().tolist()

url = "https://api.unipept.ugent.be/api/v1/pept2lca"
response = requests.post(url, data={
    'input[]': peptide_list,
    'equate_il': 'true',
    'extra': 'true'
})

taxa = response.json()
taxa_df = pd.DataFrame(taxa)

# Show organism distribution
print("\nTaxonomic Classification")
if 'taxon_name' in taxa_df.columns:
    organism_counts = taxa_df['taxon_name'].value_counts().head(25)
    print(organism_counts)
else:
    print("No taxonomic matches")
```

## Determine Family From Diverse Taxonomy

All the taxonomic counts are for unique peptide matches. Unipept matches by lowest common ancestor, thus, uncertain peptides are higher in the taxonomic status, certain peptides are lower.

To find the family of which these peptides are derived we can do the following:

- Strip all taxonomies above Family
- For all taxonomies below family, find their family and reassign the counts to the parent family.

- The family with the largest quantity of peptides represents the confidence score of unipepts matching (due to the LCA being lower) and the abundance of matches.

```python

def collapse_to_family(tax_counts_df, taxon_col="taxon_name", count_col="counts"):

    ncbi = NCBITaxa()

    # Ranks to exclude (above family)
    above_family = {
        "no rank", "root", "superkingdom", "kingdom", "subkingdom",
        "phylum", "subphylum", "class", "subclass", "order",
        "suborder", "infraorder", "superfamily"
    }

    name_to_family_cache = {}

    def get_family(taxon_name):
        if pd.isna(taxon_name):
            return None

        taxon_name = str(taxon_name).strip()

        if taxon_name in name_to_family_cache:
            return name_to_family_cache[taxon_name]

        # Skip common junk names
        if taxon_name.lower() in {"root", "unclassified", "unknown", "other"}:
            name_to_family_cache[taxon_name] = None
            return None

        try:
            # Get taxon ID
            taxid_map = ncbi.get_name_translator([taxon_name])
            taxids = taxid_map.get(taxon_name, [])
            if not taxids:
                name_to_family_cache[taxon_name] = None
                return None

            taxid = taxids[0]
            rank = ncbi.get_rank([taxid]).get(taxid, None)

            # Skip if above family level
            if rank in above_family:
                name_to_family_cache[taxon_name] = None
                return None

            # Find family in lineage
            lineage = ncbi.get_lineage(taxid)
            ranks = ncbi.get_rank(lineage)
            names = ncbi.get_taxid_translator(lineage)

            family_taxids = [t for t in lineage if ranks.get(t) == "family"]
            if not family_taxids:
                name_to_family_cache[taxon_name] = None
                return None

            family_name = names[family_taxids[0]]
            name_to_family_cache[taxon_name] = family_name
            return family_name

        except:
            name_to_family_cache[taxon_name] = None
            return None

    # Map each taxon to its family
    temp_df = tax_counts_df[[taxon_col, count_col]].copy()
    temp_df["family"] = temp_df[taxon_col].map(get_family)
    temp_df = temp_df.dropna(subset=["family"])

    # Sum counts by family
    family_df = (
        temp_df.groupby("family", as_index=False)[count_col]
        .sum()
        .sort_values(count_col, ascending=False)
        .reset_index(drop=True)
    )

    return family_df

organism_counts_df = organism_counts.reset_index()
organism_counts_df.columns = ['taxon_name', 'counts']

family_counts = collapse_to_family(organism_counts_df,
                                   taxon_col="taxon_name",
                                   count_col="counts")
```
