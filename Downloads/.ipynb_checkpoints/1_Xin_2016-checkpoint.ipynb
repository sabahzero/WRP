````r
library(tidyverse)
library(ggpubr)
````

````r
base.dir <- getwd()
````

````r
###----- Merge files & create rpkm df
count.dir <- file.path(base.dir, "1_Xin_2016", "GSE68056_RAW")
count.files <- list.files(path=count.dir, pattern=".txt", full.names=TRUE)

all.log2rpkm.df <- read_tsv(count.files[1]) %>% select(one_of(c('Symbol'))) %>% drop_na() %>% distinct()
for (i in count.files){
    i.name <- gsub(".txt","",tail(unlist(strsplit(i, "/")), 1))
    i.name <- unlist(strsplit(i.name, "_"))
    i.name <- paste(i.name[3:length(i.name)], collapse="_")
    
    i.df <- read_tsv(i) %>% 
        select(one_of(c('Symbol','log2_RPKM'))) %>% 
        drop_na(Symbol) %>% distinct(Symbol, .keep_all=TRUE)
    colnames(i.df) <- c('Symbol', i.name)
    all.log2rpkm.df <- all.log2rpkm.df %>% left_join(i.df, by='Symbol')
    
}

all.rpkm.df <- all.log2rpkm.df %>% mutate_if(is.numeric, function(x) 2**x)
                                             
gene.names <- all.rpkm.df$Symbol
                                             
all.rpkm.df <- all.rpkm.df %>% column_to_rownames(var='Symbol')
                                             
###---- Calculate z score
scaled.df <- data.frame(scale(t(all.rpkm.df)))
scaled.df$condition <- c( rep("WT", 8), rep("Blimp1KO", 2), rep("CD25KO", 2), 
                         rep("CD25KO_Blimp1KO", 2), rep("TbetKO", 2))

mean.scaled.df <- scaled.df %>% group_by(condition) %>% summarize_all(list(mean))
````

````r
plot.genes <- c("Gzma", "Cx3cr1", "Sema4a", "Ccr5", "Il7r", "Il17a","Batf3")
plot.df <- mean.scaled.df %>% select(one_of(c("condition", plot.genes))) %>% 
    pivot_longer(cols=plot.genes, names_to="gene_names")
head(plot.df)
````

````r
ggplot(plot.df, aes(condition, gene_names,fill=value)) +
    geom_tile() +
    scale_fill_viridis_c()
````

## Use pre-combined matrix

````r
log2rpkm.file <- file.path(base.dir, "1_Xin_2016", "PMID-26950239-8Lib-log2RPKM.csv")

###----- Data preprocessing
# Cleaning
all.log2rpkm.df <- read_csv(log2rpkm.file) %>% drop_na(Symbol) %>% distinct(Symbol, .keep_all=TRUE) 
# log2 to rpkm
all.rpkm.df <- all.log2rpkm.df %>% mutate_if(is.numeric, function(x) 2**x)                                         
# Convert symbol to colname for scaling (leave only numeric matrix)                                         
all.rpkm.df <- all.rpkm.df %>% column_to_rownames(var='Symbol')
                                             
###---- Calculate average z score
# Z score
scaled.df <- data.frame(scale(t(all.rpkm.df))) 
scaled.df$condition <- c( rep("WT", 2), rep("Blimp1KO", 2), rep("TbetKO", 2), rep("DKO", 2))
# Averging
mean.scaled.df <- scaled.df %>% group_by(condition) %>% summarize_all(list(mean))
````

### Figure 7A

````r
# We could provide this gene list from students
plot.genes <- c("Gzma", "Cx3cr1", "Sema4a", "Ccr5", "Prf1", "Aqp9", "Cercam", "Zeb2", "Dok2", "Klrg1", 
                "S1pr5", "Il18rap", "Gzmk", "Serpinb9", "Klrk1", "Tbx21", "Zfp683", "Ctla2a", "Gzmm", 
                "Il12rb2", "Gzmb", "Havcr2", "Fasl", "Id2", "Ccr2", "Ifng", "Cxcr6", "Cish", "Il7r", 
                "Tnfrsf9", "Ccr6", "Egr1", "Egr2", "Lag3", "Itgae", "Icos", "Hif1a", "Tspan32", "Tnfrsf4", 
                "Batf", "Bcl6", "Btla", "Pou2af1", "Irf4", "Il1r2", "Ltb", "Tnf", "Il23r", "Ccr7", "Cxcr5", 
                "Id3", "Lta", "Tox", "Tcf7", "Batf3", "Maf", "Ccr9", "Il17a", "Rorc")
conditions <- c('WT', "Blimp1KO", "TbetKO", "DKO")

# Select genes & convert wide dataframe to long form for plotting
plot.df <- mean.scaled.df %>% select(one_of(c("condition", plot.genes))) %>% 
    pivot_longer(cols=plot.genes, names_to="gene_names")

# Factorization with levels for ordered plot
plot.df$gene_names <- factor(plot.df$gene_names, levels=rev(plot.genes))
plot.df$condition <- factor(plot.df$condition, levels=conditions)
````

````r
# Ggplot heatmap
ggplot(plot.df, aes(condition, gene_names,fill=value)) +
    geom_tile() +
    scale_fill_viridis_c(option="E") + # Color scheme
    scale_y_discrete(position="right") + # Adjust y axis to the right
    theme(panel.background = element_blank())
````

### Figure 7C

````r
# Calculate average rpkm for each condition
all.rpkm.df.t <- data.frame(t(all.rpkm.df))
all.rpkm.df.t$condition <- c( rep("WT", 2), rep("Blimp1KO", 2), rep("TbetKO", 2), rep("DKO", 2))
all.rpkm.mean.df <- all.rpkm.df.t %>% group_by(condition) %>% summarize_all(list(mean))
````

````r
# Store plots in list
barplot.list <- list()
plot.genes <- c("Gzma", "Gzmk", "Gzmb", "Gzmc", "Gzmm", "Prf1") # Genes for plotting
conditions <- c('WT', "Blimp1KO", "TbetKO", "DKO")

for (plot.gene in plot.genes) {
    gene.df <- all.rpkm.mean.df %>% select(one_of(c("condition", plot.gene))) # Select genes for plotting
    gene.df$condition <- factor(gene.df$condition, levels=conditions) # Factorization for the order
    colnames(gene.df) <- c('condition', 'gene_name')
    # Plot
    barplot.list[[plot.gene]] <- ggplot(gene.df) +
    geom_bar(aes(x=condition, y=gene_name, fill=condition), color="black", stat='identity', width=0.6) +
    scale_fill_manual(values=c("black", "gray30", "gray60", "white"), guide=FALSE) +
    ylab(paste(plot.gene, "(RPKM)")) +
    theme(axis.line = element_line(colour = "black"),
          panel.grid.major = element_blank(),
          panel.grid.minor = element_blank(),
          panel.background = element_blank(),
          axis.title.x=element_blank(),
          axis.text.x = element_text(angle = 45, vjust = 0.5, hjust=0.5))
}

ggarrange(plotlist=barplot.list, ncol=3, nrow=2)
````
