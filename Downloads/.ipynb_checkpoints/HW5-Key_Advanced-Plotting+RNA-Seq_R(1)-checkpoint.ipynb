```ruby
# 3
# Students should figure out that this is only representative of one experiment, 
# and they need several for statistical power (which we'll go through next)
# Use random number generator to add variation to the data --> create "replicates"

df.rep1 <- data.frame(df)
df.rep2 <- data.frame(df)
df.rep3 <- data.frame(df)

randomizer <- runif(nrow(df), min=0.9, max=1.1)
df.rep2$Count <- as.integer(df.rep2$Count * randomizer)

randomizer <- runif(nrow(df), min=0.9, max=1.1)
df.rep3$Count <- as.integer(df.rep3$Count * randomizer)

df.rep1$replicate <- rep("rep1", nrow(df.rep1))
df.rep2$replicate <- rep("rep2", nrow(df.rep2))
df.rep3$replicate <- rep("rep3", nrow(df.rep3))

df.with.rep <- rbind(df.rep1, df.rep2, df.rep3)

###----- Boxplot

df.plot.rep <- df.with.rep[df.with.rep$ENSEMBL_ID %in% c('ENSG00000000419', 'ENSG00000000457'), ]

# Tidyverse filter
df.plot.rep <- df.with.rep %>% filter(ENSEMBL_ID %in% c('ENSG00000000419', 'ENSG00000000457'))
```

```ruby
# For comparing two genes in the same dataset, paired t-test is more appropriate
# T-test assume normal distribution of population, which is actually not true for RNAseq. 
# This is why usually in DESeq2 negative binomial distribution is used for estimating population distribution.
# DESeq2 will be covered in future classes

gene1.vec <- df.plot.rep %>% filter(ENSEMBL_ID == 'ENSG00000000419') %>% arrange(replicate) %>% .$Count
gene2.vec <- df.plot.rep %>% filter(ENSEMBL_ID == 'ENSG00000000457') %>% arrange(replicate) %>% .$Count

t.test(gene1.vec, gene2.vec, paired=TRUE)
```
