````r
library(ggsignif)
#ending in 1000's-group "a"
family1 <- df %>%
    filter(str_detect(EnsembleID, "ENSG00000001"))
family1 <- cbind(family1, group="a")
#family1

#ending in 3000's-group "b"
family2 <- df %>%
    filter(str_detect(EnsembleID, "ENSG00000003"))
family2 <- cbind(family2, group="b")
#family2


data.frame <- rbind(family1,family2)
data.frame
#new family/group data

bxplot <- ggplot(data.frame, aes(x=group, y=Norm_exp_counts, fill=group)) + geom_boxplot() +
    geom_signif(comparisons=list(c("a","b")),test = "t.test", map_signif_level=TRUE)
bxplot

#generating 2 replicates for 2 genes of interest
set.seed(12)
s1 <- sample(512, size=2, replace=FALSE) #EnsembleID:ENSG00000001036	512:count
s2 <- sample(497, size=2, replace=FALSE) #EnsembleID:ENSG00000003249	2054:count
s3<- rbind(s1,s2)
rownames(s3) <- c()
colnames(s3) <- c("Rep2", "Rep3")
s3#matrix of replicates

#Combining replicates with original gene dataframe (subsetted from main df)
data_rep <- data.frame(EnsembleID=c("ENSG00000001036", "ENSG00000003249"),
                       Rep1=c(512, 2054))
data_rep <- cbind(data_rep, s3)
data_rep

data_rep_melt <- melt(data_rep)

# boxplot for 2 gene of interest
bxplot_gene <- ggplot(data_rep_melt, aes(x=EnsembleID, y=value)) + geom_boxplot()
bxplot_gene

#Statistical significance of target genes(group of genes)

#Assumption 1: Are the two samples independents?
#Yes, since I am putting them into 2 fake groups.

#Assumtion 2: Are the data from each of the 2 groups follow a normal distribution?
#we were told in class it's a normal distribution.

#Assumption 3. Do the two populations have the same variances?
var.test(Norm_exp_counts ~ group, data = data.frame)

#The p-value of F-test is p = 0.0106. It’s lesser than the significance level alpha = 0.05. 
#In conclusion, there is  significant difference between the variances of the two sets of data.
#Therefore, we can't use the classic t-test witch assume equality of the two variances.
#so we use Welch Two sample t-test (var.equal=FALSE)

t.test(Norm_exp_counts ~ group, data = data.frame, var.equal = FALSE)

#p=0.5975 > 0.05 therefore, not siginificant.

#generating 2 replicates for 2 genes of interest
set.seed(12)
s1 <- sample(512, size=2, replace=FALSE) #EnsembleID:ENSG00000001036	512:count
s2 <- sample(497, size=2, replace=FALSE) #EnsembleID:ENSG00000003249	2054:count
s3<- rbind(s1,s2)
rownames(s3) <- c()
colnames(s3) <- c("Rep2", "Rep3")
s3#matrix of replicates

#Combining replicates with original gene dataframe (subsetted from main df)
data_rep <- data.frame(EnsembleID=c("ENSG00000001036", "ENSG00000003249"),
                       Rep1=c(512, 2054))
data_rep <- cbind(data_rep, s3)
data_rep

t.test(data_rep[1,-1], data_rep[2,-1])
#p-value=0.4437 > 0.05 therefore, not significant (but these were randomly generated by me)
````

### Question 8 (1 point)
Determine the statistical significance of these two figures and and show them in your joined plot. Describe why you chose these statistics, and the specifics of tests involve. Explain what these say about the data. (Hint: to compare 2 genes of interest, you would like to calculate fold change in $log2$-scale and get $p$-value from statistical test)




````r
#arranging both figures as one (for the 2 families of genes)
par(mfrow=c(2,2))
ggarrange(bxplot, hmap, 
          labels = c("A", "B"),
          ncol = 2, nrow = 1)

#generating 5 data values for the first 20 genes
set.seed(1000)
df1 <- df[1:20,]
m2 <- matrix(runif(20*5,0.9,1.1),20,5)

#combined generated values with original data frame
m3 <- matrix(as.integer(df1$Norm_exp_counts * m2), 20,5)
df2 <- cbind(df1,m3)



#Changed column names to treatment and control
names(df2) <- c("GeneID", "treat1", "treat2", "treat3", "control1","control2", "control3")
dimnames(df2)[[1]] <- df2[,1] #rownames as GeneID's
dfnew = df2[,-1]
#dfnew

#Function created to generate P-values  with t-test
ttestdfnew <- function(dfnew, grp1, grp2) {
  x = dfnew[grp1]
  y = dfnew[grp2]
  x = as.numeric(x)
  y = as.numeric(y)  
  results = t.test(x, y)
  results$p.value
}

#Created Histogram of raw P-values
rawpvalue = apply(dfnew, 1, ttestdfnew, grp1 = c(1:3), grp2 = c(4:6))
hist(rawpvalue)

#log2fold-change

dfnew_log2 = log2(dfnew)
treatment = apply(dfnew_log2[,1:3], na.rm=TRUE, 1, mean)
control = apply(dfnew_log2[, 4:6], na.rm=TRUE, 1, mean) 
#class(control)
#because our data is already log2 transformed, we can take the difference between the means.  
#And this is our log2 Fold Change or log2 Ratio == log2(control / test)
foldchange <- control - treatment

hist(foldchange, xlab = "log2 Fold Change (Control vs Treatment)")

#Transform the p-value (-1*log(p-value)) and create a volcano plot 
results = cbind(foldchange, rawpvalue)
results = as.data.frame(results)

results$probename <- rownames(results)

volcano = ggplot(data = results, aes(x = foldchange, y = -1*log10(rawpvalue)))
volcano + geom_point()
````

### Question 9 (1 point)
Create a loop to repeat same analysis process for multiple pairs of genes. Generate figures above with statistics as png files.


````r
n=5 #First 5 genes taken (we can change this as per required)
data.loop <- df[1:n,] 
data.loop
for (i in 1:(n-1)){  #loop within a loop to get all pairs of genes
    for (j in (i+1):n){
        s1 <- data.loop[i,]
        s2 <- data.loop[j,]
        s3<- rbind(s1,s2)
        
        #print(s1) #create replicates for each pair
         s4 <- sample(s1$Norm_exp_counts, size=2, replace=FALSE) 
         s5 <- sample(s2$Norm_exp_counts, size=2, replace=FALSE)
         s6<- rbind(s4,s5)
          rownames(s6) <- c()
        colnames(s6) <- c("Rep2", "Rep3")
        data_rep <- cbind(s3, s6)
        #print(data_rep)
        data_rep_melt <- melt(data_rep)
        #print(data_rep_melt)
        
        #plot boxplot
        bxplot_gene <- ggplot(data_rep_melt, aes(x=EnsembleID, y=value)) + geom_boxplot()+
        geom_signif(comparisons=list(c(paste(s1$EnsembleID), paste(s2$EnsembleID))),test = "t.test", map_signif_level=TRUE)
        #print(bxplot_gene)
        
        #plot heatmap
        hmap_gene <- ggplot(data_rep_melt, aes(EnsembleID, variable)) + 
        geom_tile(aes(fill=value))
        #print(hmap_gene)
        joined <- ggarrange(bxplot_gene, hmap_gene, 
        labels = c("A", "B"),
        ncol = 2, nrow = 1)
        print(joined)
        #name <- paste(s1$EnsembleID, s2$EnsembleID, sep="_")
        name <- paste("~/",s1$EnsembleID, "_", s2$EnsembleID,sep="")
        
        #Saving as png files in directory
        #ggsave(myplot,filename=paste(name,".png",sep=""))
        ggsave(filename=paste(name,".png"), width = 9, height = 5, dpi = 100, plot=joined)
        #ggsave('~/saved.png', width = 9, height = 5, dpi = 100, plot=joined)
    }
}
````
